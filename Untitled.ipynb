```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulamos el dataset con las variables mencionadas en tu TEG
np.random.seed(42)
n = 1500
df_eda = pd.DataFrame({
    'Meses_Vigencia': np.random.randint(1, 120, n),
    'Cuotas_Vencidas': np.random.choice([0, 1, 2, 3], n, p=[0.7, 0.15, 0.1, 0.05]),
    'Prima_Anual': np.random.uniform(15, 80, n),
    'Ano_Vehiculo': np.random.randint(2010, 2024, n),
    'Fuga': np.random.choice([0, 1], n, p=[0.78, 0.22]) # 22% de churn
})
# Añadimos correlación artificial para que el gráfico tenga sentido
df_eda.loc[df_eda['Fuga'] == 1, 'Cuotas_Vencidas'] += 1
df_eda.loc[df_eda['Fuga'] == 1, 'Meses_Vigencia'] = np.random.randint(1, 24, sum(df_eda['Fuga'] == 1))

sns.set_theme(style="whitegrid")

# --- GRÁFICO 1: Matriz de Correlación ---
plt.figure(figsize=(8, 6))
correlation_matrix = df_eda.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Figura X: Matriz de Correlación de Variables Lineales', fontweight='bold', pad=15)
plt.show()

# --- GRÁFICO 2: Boxplot de Antigüedad vs Fuga ---
plt.figure(figsize=(7, 5))
sns.boxplot(x='Fuga', y='Meses_Vigencia', data=df_eda, palette=['#28a745', '#dc3545'])
plt.xticks([0, 1], ['Retenidos (0)', 'Fuga (1)'])
plt.ylabel('Meses de Vigencia de la Póliza')
plt.title('Figura Y: Distribución de la Antigüedad por Estado del Cliente', fontweight='bold')
plt.show()

# --- GRÁFICO 3: Método de Pago vs Fuga ---
# Simulamos datos categóricos
metodos = ['PAT (Cuenta Vista)', 'PAT (Tarjeta Crédito)', 'PAC (Cuenta Corriente)']
df_pagos = pd.DataFrame({
    'Metodo': np.random.choice(metodos, n, p=[0.4, 0.4, 0.2]),
    'Fuga': df_eda['Fuga']
})
plt.figure(figsize=(8, 5))
sns.countplot(x='Metodo', hue='Fuga', data=df_pagos, palette=['#28a745', '#dc3545'])
plt.title('Figura Z: Fuga de Clientes según Método de Pago', fontweight='bold')
plt.ylabel('Cantidad de Clientes')
plt.xlabel('Método de Recaudación')
plt.legend(title='Estado', labels=['Retenido', 'Fuga'])
plt.show()
```
